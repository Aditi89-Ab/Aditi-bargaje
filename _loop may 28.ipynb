```python
# find the addition of digits of a number
# eg -> 12345 => 1 + 2 + 3 + 4 + 5 = 15
```

```python
# hint -> make use of % and //
```

```python
num = int(input("enter a number:"))

add = 0
while num > 0:
    digit = num % 10
    add = add + digit
    num = num // 10
print("additon of digits =",add)    
```

```python
num = 12345

add = 0

while num > 0:
    rem = num % 10
    add = add + rem
    num = num // 10

print(add)    
```
