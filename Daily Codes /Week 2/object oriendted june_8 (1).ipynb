```python
class Employee:
    company = "ABC Technologies"      # Common for all employees

    def __init__(self, name, monthly_salary, job_designation):
        self.name = name
        self.__monthly_salary = monthly_salary   # Private variable
        self.job_designation = job_designation

    # Method to calculate annual salary
    def annual_salary(self):
        return self.__monthly_salary * 12

    # Method to display employee details
    def show_data(self):
        print("Company:", Employee.company)
        print("Name:", self.name)
        print("Job Designation:", self.job_designation)
        print("Monthly Salary:", self.__monthly_salary)
        print("Annual Salary:", self.annual_salary())
       
```

```python
e1 = Employee("Aarav", 30000, "Software Engineer")
e2 = Employee("Priya", 40000, "Data Analyst")
e3 = Employee("Karan", 50000, "Project Manager")
```

```python
e1.show_data()
e2.show_data()
e3.show_data()
```
