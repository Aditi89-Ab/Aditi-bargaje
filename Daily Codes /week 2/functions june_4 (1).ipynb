```python
# create a function to find th factorial of a number
```

```python
def fact(n):
    f = 1
    for i in range (1,n+1):
        f = f * i
    print(f)    
```

```python
fact(5)
```

```python
fact(10)
```

```python
# pass any number of integers in a function and find their sum
```

```python
def find_sum(*numbers):
    total = 0
    for num in numbers:
        total += num
    return total
```

```python
print(find_sum(10, 20, 30))      
```
