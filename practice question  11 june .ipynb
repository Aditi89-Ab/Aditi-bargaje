```python
# 2) sum of even numbers
```

```python
numbers = [12,7,23,48,15,66,9]

sum_even = 0

for num in numbers:
  if num % 2 == 0:
      sum_even += num

print("sum of even numbers",sum_even)
```

```python
#  3) manual list reversal
```

```python
data = ['apple','banana','cherry','date','elderberry']

reverse_list = []

for i in range(len(data)-1,-1,-1):
    reverse_list.append(data[i])
print(reverse_list)    
```

```python
# 4)  finding the maximum element
```

```python
values = [8, 2, 10, 5, 17, 3, 14]

largest = values[0]

for num in values:
    if num > largest:
        largest = num

print("Largest number:", largest)
```

```python
# 7)  list palindrome check
```

```python
list1 = [1, 2, 3, 2, 1]

palindrome = True

for i in range(len(list1) // 2):
    if list1[i] != list1[len(list1) - 1 - i]:
        palindrome = False
        break

if palindrome:
    print("Palindrome")
else:
    print("Not a Palindrome")
```

```python
# 8) merge and sort two lists (manual sort)
```

```python
list_a = [3, 1, 4]
list_b = [2, 5, 0]

merged = list_a + list_b

for i in range(len(merged)):
    for j in range(len(merged) - 1 - i):
        if merged[j] > merged[j + 1]:
            merged[j], merged[j + 1] = merged[j + 1], merged[j]

print(merged)
```

```python
# 9) flatten a nested list
```

```python
nested_list = [1, [2, 3], 4, [5, 6, 7], 8]

flat_list = []

for item in nested_list:
    if type(item) == list:
        for x in item:
            flat_list.append(x)
    else:
        flat_list.append(item)

print(flat_list)
```
